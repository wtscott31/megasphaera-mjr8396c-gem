# MJR8396C v6: actual results for the four requested workstreams
This executed review reads saved scientific outputs. New exact-strain reconstructions are not claimed as recovered published models; genomic and computational evidence is not an enzyme or enantiomer assay.

```python
from pathlib import Path
import json,hashlib,sys
import numpy as np,pandas as pd
ROOT=Path.cwd()
def j(p):return json.loads((ROOT/p).read_text())
print('Actual interpreter:',sys.executable)
facts=j('results/reviewer_v6/facts.json')
assert facts['native_community_cases']==1920 and facts['native_community_optimal_cases']==1920
display(pd.read_csv(ROOT/'TASK_STATUS_v6.csv'))
```

## Full new matched NM10/BL7 suite

```python
a=pd.read_csv(ROOT/'tables/S74_new_matched_reconstruction_audit_v6.csv')
assert len(a)==4
display(a[['model','reactions','metabolites','genes','added_reactions']])
display(pd.read_csv(ROOT/'results/reviewer_v6/matched_media_summary.csv'))
aa=pd.read_csv(ROOT/'tables/S76_matched_AA_singles_pairs_v6.csv');assert len(aa)==840
loop=pd.read_csv(ROOT/'tables/S79b_matched_native_loopless_FVA_v6.csv');assert len(loop)==81 and loop.status.eq('optimal').all()
print('AA records:',len(aa),'accepted loopless FVA ranges:',len(loop))
```

```python
g=pd.read_csv(ROOT/'tables/S82_primary_GPR_curation_v6.csv');assert len(g)==3
display(g[['reaction','before_GPR','after_GPR','decision']])
display(j('results/matched_functional_curation_v6/NM10_Bcd_all_feature_check.json'))
x=pd.read_csv(ROOT/'tables/S78_matched_interventions_v6.csv')
y=x[x.model.isin(['MJR8396C','NM10','BL7'])&x.design.isin(['OK2','OK3','OK4'])&x.level.eq('gene')]
assert len(y)==18 and y.growth.abs().max()<1e-7
print('All18 tested OK2-OK4 global gene implementations remain non-growing.')
```
